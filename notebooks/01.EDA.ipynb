```python
from pathlib import Path
import pandas as pd

# Where the CSV files live (from inside the notebooks folder)
data_path = Path("../data/raw")

# An empty list to collect each app's sample
all_samples = []

# Go through the CSV files one at a time
for file in sorted(data_path.glob("*.csv")):

    # "chatgpt_reviews.csv" -> "chatgpt"
    app_name = file.stem.replace("_reviews", "")

    # Read the whole file
    df = pd.read_csv(file)

    # Pick 100,000 random rows
    sample = df.sample(n=100_000, random_state=42)

    # Remember which app this came from
    sample["app_name"] = app_name

    # Add it to our list
    all_samples.append(sample)

    print(app_name, "->", len(sample), "rows kept out of", len(df))

# Stick all five samples on top of each other into one table
master = pd.concat(all_samples, ignore_index=True)

print("Final shape:", master.shape)
```

```python
before = len(master)

# Keep the first occurrence of each reviewId, drop later duplicates.
# subset=["reviewId"] means: only look at that ONE column when deciding
# what counts as a duplicate. We never dedupe on content.
master = master.drop_duplicates(subset=["reviewId"])

after = len(master)
print("Removed", before - after, "duplicate reviewIds")
print("Shape now:", master.shape)
before = len(master)

# dropna with subset=[...] drops a row if ANY of the listed columns is empty.
# We list ONLY content and score — nothing else justifies dropping a row.
master = master.dropna(subset=["content", "score"])

after = len(master)
print("Removed", before - after, "rows missing content or score")
print("Shape now:", master.shape)
before = len(master)

# dropna with subset=[...] drops a row if ANY of the listed columns is empty.
# We list ONLY content and score — nothing else justifies dropping a row.
master = master.dropna(subset=["content", "score"])

after = len(master)
print("Removed", before - after, "rows missing content or score")
print("Shape now:", master.shape)
# Confirm the columns we promised to KEEP still have their missing values
print(master.isna().sum())

# Confirm content is never empty now
print("\nRows with empty content:", (master["content"].str.strip() == "").sum())
# Make sure the output folder exists (won't error if it already does)
Path("../data/processed").mkdir(parents=True, exist_ok=True)

# Internal working file -> Parquet (needs pyarrow, already installed)
master.to_parquet("../data/processed/master_clean.parquet", index=False)

print("Saved:", master.shape, "-> ../data/processed/master_clean.parquet")
check = pd.read_parquet("../data/processed/master_clean.parquet")
print(check.shape)
print(check.dtypes)
```
